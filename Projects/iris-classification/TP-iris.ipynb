```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv("iris.csv")
x = iris.loc[:, "petal_length"]
y = iris.loc[:, "petal_width"]
lab = iris.loc[:, "species"]
```

```python
plt.axis('equal')
plt.scatter(x[lab == 0], y[lab == 0], color='g', label='setosa')
plt.scatter(x[lab == 1], y[lab == 1], color='r', label='versicolor')
plt.scatter(x[lab == 2], y[lab == 2], color='b', label='virginica')
plt.legend()
plt.show()
```

```python
# valeurs à chercher
longueur=2
largeur=0.5

# graphique

plt.axis('equal')
plt.scatter(x[lab == 0], y[lab == 0], color='g', label='setosa')
plt.scatter(x[lab == 1], y[lab == 1], color='r', label='versicolor')
plt.scatter(x[lab == 2], y[lab == 2], color='b', label='virginica')
plt.scatter(longueur, largeur, color='k')
plt.legend()
plt.show()
```

```python
longueur=5
largeur=1.4

from sklearn.neighbors import KNeighborsClassifier
k=3
d=list(zip(x,y))
model = KNeighborsClassifier(n_neighbors=k)
model.fit(d, lab)
prediction= model.predict([[longueur, largeur]])
print(prediction)
```

```python
# Affichage des résultats

# graphique

#  test appartenance
plt.axis('equal')
plt.scatter(x[lab == 0], y[lab == 0], color='g', label='setosa')
plt.scatter(x[lab == 1], y[lab == 1], color='r', label='versicolor')
plt.scatter(x[lab == 2], y[lab == 2], color='b', label='virginica')
plt.scatter(longueur, largeur, color='k')
plt.legend()


# Attention, "predection" est une liste Python qui contient un seul élément (le label), 
# il est donc nécessaire d'écrire "predection[0]" afin d'obtenir le label.

txt="Résultat de la prédiction : "
if prediction[0]==0:
    txt=txt+ "Setosa"
if prediction[0]==1:
    txt=txt+ "versicolor"
if prediction[0]==2:
    txt=txt+ "virginica"

plt.text(3,0.5, f"largeur : {largeur} cm longueur : {longueur} cm", fontsize=8)
plt.text(3,0.2, f"k : {k}", fontsize=8)
plt.text(3,-0.1, txt, fontsize=8)

plt.show()
```
